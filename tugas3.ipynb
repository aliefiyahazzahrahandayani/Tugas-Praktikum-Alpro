1. Membuat data belanjaan 

```python
# Membuat list belanjaan 
belanjaan=["Apel","Susu","Roti"]
# Membuat harga_jumlah dengan tuple
harga_jumlah=(5000,3,12000,2,8000,1)
# Menampilkan karakter pertama dari barang pertama dan karakter terakhir dari barang terakhir
print(f"Karakter pertama dari barang terakhir : {belanjaan[0][0]}")
print(f"Karakter terakhir dari baranf terakhir : {belanjaan[-1][-1]}")
print("-"*50)
```

2. Hitung total harga barang

```python
# Menghitumg total harga masing-masing secara manual 
Htotal_apel = harga_jumlah[0]*harga_jumlah[1]
Htotal_susu = harga_jumlah[2]*harga_jumlah[3]
Htotal_roti = harga_jumlah[4]*harga_jumlah[5]
# Menghitang total keselurahan dengan menjumlahkan total per barang
total_belanjaan = Htotal_apel + Htotal_susu + Htotal_roti
# Mengecek apakah hasilnya sesuai dengan print 
print(Htotal_apel,Htotal_susu,Htotal_roti)
print(total_belanjaan)
```

3. Mencetak struk belanjaan

```python
# Memenbuat variabel untuk nama toko dan pembeli
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"
# Mencetak Struk belanjaan dengan string dan escape sequences
print('='*41)
print(f'{nama_toko:^41}')
print('='*41)
print(f'Pembeli : {nama_pembeli}')
# Memberi judul pada kolom struk
print(f'{'Barang':<15}{'Harga':<8}{'Jumlah':<8}{'Total'}')
print('-'*41)
# Isi data data di struk
print(f'{belanjaan[0]:<15}{harga_jumlah[0]:<8}{harga_jumlah[1]:<8}{Htotal_apel}')
print(f'{belanjaan[1]:<15}{harga_jumlah[2]:<8}{harga_jumlah[3]:<8}{Htotal_susu}')
print(f'{belanjaan[2]:<15}{harga_jumlah[4]:<8}{harga_jumlah[5]:<8}{Htotal_roti}')
print('-'*41)
# Menhitung total keseluruhan
print(f'{'Total Belanja' :<31}{total_belanjaan}')
print('='*41)
# mengeprint ucapan terimakasih 
print(' Terima kasih sudah belanja!', chr(0x1F64F))
print('='*41)
```
