```python
# --- MENGINPUT VARIABEL-VARIABEL YANG DIPERLUKAN ---
usia_pengemudi = int(input("Masukkan usia pemilik kendaraan (dalam tahun): "))
jenis_kendaraan = input("Masukkan tipe kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (dalam rupiah): "))
area_penggunaan = input("Masukkan lokasi penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah memiliki sistem pengaman (ya/tidak): ").lower()
# --- PENENTUAN KATEGORI BERDASARKAN RISIKO ---
# 1. Cek kondisi kendaraan terlebih dahulu
if jenis_kendaraan == "motor":
    kategori_risiko = "sedang"
else:
    kategori_risiko = "rendah"
# 2. Cek faktor usia, area, dan pengaman untuk menaikkan status ke "tinggi"
if usia_pengemudi < 20 or usia_pengemudi > 60:
    kategori_risiko = "tinggi"
elif area_penggunaan == "perkotaan" and kategori_risiko == "sedang":
    kategori_risiko = "tinggi"
elif sistem_pengaman == "tidak" and kategori_risiko == "sedang":
    kategori_risiko = "tinggi"
# Cek penyesuaian dari rendah ke sedang
if area_penggunaan == "perkotaan" and kategori_risiko == "rendah":
    kategori_risiko = "sedang"
if sistem_pengaman == "tidak" and kategori_risiko == "rendah":
    kategori_risiko = "sedang"
# --- PERHITUNGAN PREMI ASURANSI ---
if kategori_risiko == "tinggi":
      persentase_premi = 0.1
elif kategori_risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05
# --- Menghitung premi yang akan dibayar
premi_dibayar = persentase_premi * nilai_kendaraan
# --- PESAN DARI DORAEMON UNTUK PENGGUNA APLIKAI ---
# Conditional Expression (Satu Baris) seperti yang dimateri link ituu
pesan = "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!" if kategori_risiko == "tinggi" else "Risiko kendaraan masih aman menurut Doraemon"
# --- OUTPUT PROGRAM ---
print("=== OUTPUT ===")
print(f"1. Kategori risiko: {kategori_risiko}")
print(f"2. Premi asuransi yang perlu dibayarkan: Rp {premi_dibayar}")
print(f"3. Pesan Doraemon: {pesan}")
```
