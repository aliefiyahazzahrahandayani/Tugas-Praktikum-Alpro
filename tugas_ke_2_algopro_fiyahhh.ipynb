1) Menyimpan Data Diri Giant

```python
# Simpan data diri Giant dalam variabel
Nama = "Giant"
Usia = 20
Berat_Badan = "97.5"
Tinggi_Badan = 1.75
Status_olahraga = False

# Cetak data diri Giant
print("Nama:",Nama)
print("Usia:",Usia,"Tahun")
print("Berat Badan:",Berat_Badan,"Kg")
print("Tinggi Badan:",Tinggi_Badan,"m")
print("Status Olahraga:",Status_olahraga)
```

2. Mengubah variabel berat badan yang tadinya string menjadi float

```python
berat_badan = float(Berat_Badan)
print("Halo, saya",Nama,"Usia saya",Usia,"tahun dan Berat badan saya",berat_badan,"kg")
```

3. Menghitung BMI Giant lalu hasilnya di bulatkan ke 2 desimal

```python
# Menghitung berat badan standar Giant
BMI = berat_badan / (Tinggi_Badan ** 2)
print("BMI Giant adalah",round(BMI,2))
```

4. Perhitungan target minum air selama diet

```python
target_air = 2000 #dalam ml
kapasitas_botol = 350 # dalam ml
# Menggunakan true division untuk hasil asli
hasil_asli_botol = target_air/kapasitas_botol
# Menggunakan floor division untuk botol penuh
botol_penuh = target_air//kapasitas_botol
# Menggunakan modulo untuk sisa air
sisa_air = target_air%kapasitas_botol
# Mencetak hasil perhitungan air
print("Hasil asli pembagi botol:",hasil_asli_botol)
print("Jumlah botol penuh:",botol_penuh,"botol")
print("Jumlah air yang tersisa:",sisa_air,"ml")
```

5. Perhitungan pembulatan kalori harian

```python
import math
# Menyimpan data kebutuhan kalori per hari
kalori_giant = 2100.5
# Membulatkan ke bawah menggunakan math.floor()
kalori_bulat_bawah = math.floor(kalori_giant)
# Membulatkan ke atas menggunakan math.ceil()
kalori_bulat_atas = math.ceil(kalori_giant)
# Mencetak hasil pembulatan kaalori
print("Kalori Giant dibulatkan ke bawah:",kalori_bulat_bawah,"kalori")
print("Kalori Giant dibulatkan ke atas:",kalori_bulat_atas,"kalori")
```
